## Mục lục

4. Đưa ra giải pháp

```python
import pandas as pd
import numpy as np

# Đọc dữ liệu
df = pd.read_csv("googleplaystore.csv")
print(f"Kích thước: {df.shape[0]} dòng x {df.shape[1]} cột")
df.head()
```

## 1. Chọn Dataset

- **Tên:** Google Play Store Apps
- **Nguồn:** Kaggle (Lavanya Gupta)
- **Mô tả:** Chứa thông tin các ứng dụng trên Google Play Store
- **Số dòng:** ~10,841
- **Số cột:** 13

## 2. Dataset có gì?

```python
print("=== Tên cột và kiểu dữ liệu ===")
print(df.dtypes)
print("\n=== Thông tin tổng quan ===")
df.info()
```

```python
print("=== 5 dòng đầu ===")
display(df.head())
print("\n=== 5 dòng cuối ===")
display(df.tail())
```

```python
print("=== Giá trị thiếu ===")
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({'Số thiếu': missing, 'Tỷ lệ %': missing_pct.round(2)})
display(missing_df[missing_df['Số thiếu'] > 0])
```

```python
print("=== Dòng trùng lặp ===")
print(f"Số dòng trùng: {df.duplicated().sum()}")
print(f"\nSố App trùng tên: {df['App'].duplicated().sum()}")
```

## 3. Đánh giá vấn đề

```python
print("=== Vấn đề 1: Cột Reviews chứa giá trị không phải số ===")
reviews_invalid = df[pd.to_numeric(df['Reviews'], errors='coerce').isna()]
print(f"Số dòng lỗi: {len(reviews_invalid)}")
if len(reviews_invalid) > 0:
    display(reviews_invalid.head())

print("\n=== Vấn đề 2: Cột Size có đơn vị khác nhau (M, k) ===")
print(df['Size'].apply(lambda x: str(x)[-1] if pd.notna(x) else 'NaN').value_counts())

print("\n=== Vấn đề 3: Cột Installs có ký tự đặc biệt (+, ,) ===")
print(df['Installs'].unique()[:10])

print("\n=== Vấn đề 4: Cột Price có ký tự $ ===")
print(df['Price'].apply(lambda x: str(x)[0] if pd.notna(x) else 'NaN').value_counts())

print("\n=== Vấn đề 5: Cột Last Updated là chuỗi ===")
print(df['Last Updated'].head(3))

print("\n=== Vấn đề 6: Rating ngoài phạm vi (0-5) ===")
print(df['Rating'].describe())
invalid_rating = df[(df['Rating'] < 0) | (df['Rating'] > 5)]
print(f"Số dòng Rating ngoài [0, 5]: {len(invalid_rating)}")
if len(invalid_rating) > 0:
    display(invalid_rating)
```

```python
print("=== Vấn đề 7: Genres có dấu ; (multi-category) ===")
multi_genre = df[df['Genres'].str.contains(';', na=False)]
print(f"Số dòng có nhiều Genres: {len(multi_genre)}")

print("\n=== Vấn đề 8: Current Ver và Android Ver có giá trị 'Varies with device' ===")
print(f"Current Ver 'Varies with device': {(df['Current Ver'] == 'Varies with device').sum()}")
print(f"Android Ver 'Varies with device': {(df['Android Ver'] == 'Varies with device').sum()}")
```

## 4. Giải pháp

```python
# Tạo bản sao để xử lý
df_clean = df.copy()
print(f"Trước clean: {df_clean.shape}")
```

```python
# 4.1: Xóa dòng trùng lặp hoàn toàn
df_clean = df_clean.drop_duplicates()
print(f"Sau xóa trùng: {df_clean.shape}")
```

```python
# 4.2: Sửa cột Reviews - tìm và xóa dòng không phải số
reviews_invalid = pd.to_numeric(df_clean['Reviews'], errors='coerce').isna()
print(f"Số dòng Reviews lỗi: {reviews_invalid.sum()}")
df_clean = df_clean[~reviews_invalid].copy()
df_clean['Reviews'] = df_clean['Reviews'].astype(int)
```

```python
# 4.3: Xóa dòng có Rating ngoài [0, 5]
df_clean = df_clean[(df_clean['Rating'] >= 0) & (df_clean['Rating'] <= 5)]
print(f"Sau lọc Rating: {df_clean.shape}")
```

```python
# 4.4: Chuẩn hóa cột Size (M -> số, k -> /1024)
def convert_size(val):
    if pd.isna(val):
        return np.nan
    val = str(val).strip()
    if val == 'Varies with device':
        return np.nan
    if val.endswith('M'):
        return float(val[:-1])
    if val.endswith('k'):
        return float(val[:-1]) / 1024
    return np.nan

df_clean['Size_MB'] = df_clean['Size'].apply(convert_size)
print(f"Size hợp lệ: {df_clean['Size_MB'].notna().sum()}/{len(df_clean)}")
```

```python
# 4.5: Chuẩn hóa Installs (bỏ +, ,)
df_clean['Installs_Num'] = (df_clean['Installs']
    .str.replace('+', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(int))
print(df_clean['Installs_Num'].describe())
```

```python
# 4.6: Chuẩn hóa Price (bỏ $)
df_clean['Price_Num'] = (df_clean['Price']
    .str.replace('$', '', regex=False)
    .astype(float))
print(df_clean['Price_Num'].describe())
```

```python
# 4.7: Chuyển Last Updated sang datetime
df_clean['Last_Updated_Date'] = pd.to_datetime(df_clean['Last Updated'], errors='coerce')
print(df_clean['Last_Updated_Date'].head())
```

```python
# 4.8: Điền giá trị thiếu cho Rating bằng trung bình theo Category
df_clean['Rating'] = df_clean.groupby('Category')['Rating'].transform(
    lambda x: x.fillna(x.mean())
)
print(f"Rating còn thiếu: {df_clean['Rating'].isna().sum()}")
```

```python
# 4.9: Tách Genres chính (lấy phần đầu trước dấu ;)
df_clean['Main_Genre'] = df_clean['Genres'].str.split(';').str[0]
print(df_clean['Main_Genre'].value_counts().head())
```

```python
# 4.10: Chuẩn hóa chuỗi (xóa khoảng trắng thừa)
for col in df_clean.select_dtypes(include=['object', 'str']).columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()

print(f"\nKích thước sau clean: {df_clean.shape}")
```

## 5. Kiểm chứng

```python
print("=== KIỂM CHỨNG SAU CLEAN ===")

# 5.1: Không còn dòng trùng
print(f"\n1. Dòng trùng: {df_clean.duplicated().sum()} (yêu cầu: 0)")

# 5.2: Reviews là số nguyên
print(f"2. Reviews kiểu int: {df_clean['Reviews'].dtype}")

# 5.3: Rating trong [0, 5]
print(f"3. Rating range: [{df_clean['Rating'].min():.2f}, {df_clean['Rating'].max():.2f}]")

# 5.4: Installs_Num là số nguyên
print(f"4. Installs_Num kiểu: {df_clean['Installs_Num'].dtype}")
print(f"   Installs_Num min/max: {df_clean['Installs_Num'].min()}/{df_clean['Installs_Num'].max()}")

# 5.5: Price_Num là số
print(f"5. Price_Num kiểu: {df_clean['Price_Num'].dtype}")
print(f"   Có giá trị âm: {(df_clean['Price_Num'] < 0).sum()}")

# 5.6: Last_Updated_Date là datetime
print(f"6. Last_Updated_Date kiểu: {df_clean['Last_Updated_Date'].dtype}")

# 5.7: Size_MB là số
print(f"7. Size_MB kiểu: {df_clean['Size_MB'].dtype}")
print(f"   Size_MB range: [{df_clean['Size_MB'].min():.2f}, {df_clean['Size_MB'].max():.2f}]")
```

```python
print("=== Giá trị thiếu còn lại ===")
missing_final = df_clean.isnull().sum()
print(missing_final[missing_final > 0])
```

```python
# 5.8: So sánh trước/sau
print("=== SO SÁNH TRƯỚC/SAU ===")
print(f"{'Chỉ số':<25}{'Trước':>15}{'Sau':>15}")
print("-" * 55)
print(f"{'Số dòng':<25}{len(df):>15}{len(df_clean):>15}")
print(f"{'Dòng trùng':<25}{df.duplicated().sum():>15}{df_clean.duplicated().sum():>15}")
print(f"{'Reviews lỗi':<25}{pd.to_numeric(df['Reviews'], errors='coerce').isna().sum():>15}{0:>15}")
print(f"{'Rating ngoài [0,5]':<25}{len(df[(df['Rating']<0)|(df['Rating']>5)]):>15}{len(df_clean[(df_clean['Rating']<0)|(df_clean['Rating']>5)]):>15}")
```

```python
# Lưu file đã clean
df_clean.to_csv("googleplaystore_cleaned.csv", index=False)
print("✅ Đã lưu: googleplaystore_cleaned.csv")
display(df_clean.head())
```
